## First-order method


```python
"""
Algorithm 1 experiment for the Cournot-style multiple-follower Stackelberg game.

The implementation follows the paper's three main operations:

1. Line 3: approximately solve the follower variational inequality using
   Tseng/Korpelevich projected extragradient.
2. Line 4/5: approximately minimize the surrogate Lagrangian in y using
   standard projected gradient descent.
3. Line 7: update the scalar leader variable with a projected gradient step.

Only NumPy and Matplotlib are required.
"""

from __future__ import annotations

import csv
import math
import time
from dataclasses import dataclass
from pathlib import Path
from typing import Callable, Dict, List, Optional, Sequence, Tuple

import numpy as np


Array = np.ndarray


@dataclass(frozen=True)
class CournotProblem:
    """Parameters of the scalar-leader, k-follower Cournot toy problem."""

    a: float = 12.0
    b: float = 0.05
    c: float = 2000.0
    k: int = 2000
    R_x: float = 100.0
    R_y: float = 1000.0
    mu_x: float = 10.0
    gamma: float = 1.0
    Q_des: float = 10.0

    def validate(self) -> None:
        if self.k <= 0:
            raise ValueError("k must be positive.")
        if self.b < 0 or self.c <= 0:
            raise ValueError("Require b >= 0 and c > 0.")
        if self.R_x <= 0 or self.R_y <= 0:
            raise ValueError("R_x and R_y must be positive.")
        if self.mu_x <= 0 or self.gamma <= 0:
            raise ValueError("mu_x and gamma must be positive.")

    @property
    def follower_strong_monotonicity(self) -> float:
        # Jacobian of V is (c+b)I + b 11^T.
        return self.c + self.b

    @property
    def follower_lipschitz_constant(self) -> float:
        # Largest eigenvalue of (c+b)I + b 11^T.
        return self.c + self.b * (self.k + 1)

    @property
    def symmetric_response_denominator(self) -> float:
        # Interior symmetric equilibrium satisfies
        # [c + b(k+1)] y_i = a + x.
        return self.follower_lipschitz_constant

    @property
    def interior_upper_smoothness(self) -> float:
        # On the region where the follower equilibrium is interior,
        # F''(x) = mu_x + gamma k^2 / d^2.
        d = self.symmetric_response_denominator
        return self.mu_x + self.gamma * self.k**2 / d**2


@dataclass(frozen=True)
class AlgorithmConfig:
    """Numerical choices for Algorithm 1."""

    T: int = 600
    x0: float = 0.0
    y0_value: float = 0.0
    z0_value: float = 0.0

    # lambda_t = lambda0 * sqrt(t+1)
    lambda0: float = 1.0

    # alpha_z = extragradient_fraction / L_V, with fraction in (0,1).
    extragradient_fraction: float = 0.9

    # If eta_x is None, eta_x_fraction / L_F is used.
    # A fraction below one is useful with inexact inner solves.
    eta_x: Optional[float] = None
    eta_x_fraction: float = 0.5

    # M_z,t and M_y,t grow logarithmically.
    mz_base: int = 5
    mz_log_coefficient: float = 1.0
    my_base: int = 10
    my_log_coefficient: float = 2.0

    # Optional fixed inner iteration counts. When not None, they override
    # the logarithmic schedules.
    fixed_Mz: Optional[int] = None
    fixed_My: Optional[int] = None

    # Time-to-convergence is the cumulative runtime until the first
    # iterate with |F(x_t)-F(x*)| <= convergence_gap_tol.
    convergence_gap_tol: float = 1e-3

    def validate(self) -> None:
        if self.T <= 0:
            raise ValueError("T must be positive.")
        if self.lambda0 <= 0:
            raise ValueError("lambda0 must be positive.")
        if not 0.0 < self.extragradient_fraction < 1.0:
            raise ValueError("extragradient_fraction must lie in (0,1).")
        if self.eta_x is not None and self.eta_x <= 0:
            raise ValueError("eta_x must be positive.")
        if self.eta_x_fraction <= 0:
            raise ValueError("eta_x_fraction must be positive.")
        if self.convergence_gap_tol <= 0:
            raise ValueError("convergence_gap_tol must be positive.")
        for name, value in (
            ("mz_base", self.mz_base),
            ("my_base", self.my_base),
        ):
            if value < 0:
                raise ValueError(f"{name} must be nonnegative.")
        for name, value in (
            ("fixed_Mz", self.fixed_Mz),
            ("fixed_My", self.fixed_My),
        ):
            if value is not None and value < 0:
                raise ValueError(f"{name} must be nonnegative.")


# ---------------------------------------------------------------------------
# Projection and model primitives
# ---------------------------------------------------------------------------

def project_x(x: float, problem: CournotProblem) -> float:
    """Euclidean projection onto X = [-R_x, R_x]."""
    return float(np.clip(x, -problem.R_x, problem.R_x))


def project_y(y: Array, problem: CournotProblem) -> Array:
    """Coordinatewise Euclidean projection onto Y = [0,R_y]^k."""
    return np.clip(np.asarray(y, dtype=float), 0.0, problem.R_y)


def follower_costs(x: float, y: Array, problem: CournotProblem) -> Array:
    """Return the vector (g_1(x,y),...,g_k(x,y))."""
    y = np.asarray(y, dtype=float)
    Q = float(np.sum(y))
    return 0.5 * problem.c * y**2 - (problem.a + x - problem.b * Q) * y


def follower_operator(x: float, y: Array, problem: CournotProblem) -> Array:
    r"""
    Return V(x,y) = (partial g_i / partial y_i)_i.

    Since
        g_i = c/2 y_i^2 - (a+x-bQ)y_i,
    and Q includes y_i,
        V_i = c y_i - (a+x) + b(Q+y_i)
            = (c+b)y_i + bQ - (a+x).
    """
    y = np.asarray(y, dtype=float)
    Q = float(np.sum(y))
    return (problem.c + problem.b) * y + problem.b * Q - (problem.a + x)


def exact_follower_equilibrium(x: float, problem: CournotProblem) -> Array:
    r"""
    Closed-form Nash equilibrium of the homogeneous follower game.

    Symmetry and uniqueness imply y_1^*=...=y_k^*=s(x), where
        s(x) = clip((a+x)/(c+b(k+1)), 0, R_y).
    """
    d = problem.symmetric_response_denominator
    s = float(np.clip((problem.a + x) / d, 0.0, problem.R_y))
    return np.full(problem.k, s, dtype=float)


def leader_cost(x: float, y: Array, problem: CournotProblem) -> float:
    """Leader objective f(x,y)."""
    Q = float(np.sum(y))
    return (
        0.5 * problem.mu_x * x**2
        + 0.5 * problem.gamma * (Q - problem.Q_des) ** 2
    )


def upper_objective(x: float, problem: CournotProblem) -> float:
    """True bilevel objective F(x)=f(x,y^*(x))."""
    return leader_cost(x, exact_follower_equilibrium(x, problem), problem)


def upper_gradient(x: float, problem: CournotProblem) -> float:
    r"""
    Derivative of F on a smooth piece of the response map.

    At the two clipping breakpoints, F need not be differentiable. The code
    returns the derivative associated with the currently active clipped piece.
    The default experiment remains in the interior region.
    """
    d = problem.symmetric_response_denominator
    raw_s = (problem.a + x) / d
    if 0.0 < raw_s < problem.R_y:
        ds_dx = 1.0 / d
        s = raw_s
    else:
        ds_dx = 0.0
        s = float(np.clip(raw_s, 0.0, problem.R_y))

    Q = problem.k * s
    return (
        problem.mu_x * x
        + problem.gamma * (Q - problem.Q_des) * problem.k * ds_dx
    )


def exact_stackelberg_solution(problem: CournotProblem) -> Tuple[float, float]:
    r"""
    Compute a global minimizer of the one-dimensional piecewise-quadratic F.

    Candidate points consist of:
    - the two leader bounds;
    - follower-response clipping breakpoints;
    - the stationary point in each smooth piece.
    """
    d = problem.symmetric_response_denominator
    lower_kink = -problem.a
    upper_kink = d * problem.R_y - problem.a

    candidates: List[float] = [-problem.R_x, problem.R_x]

    for kink in (lower_kink, upper_kink):
        if -problem.R_x <= kink <= problem.R_x:
            candidates.append(float(kink))

    # Stationary point in the lower clipped region, where y^*(x)=0.
    if -problem.R_x <= 0.0 <= problem.R_x and 0.0 <= lower_kink:
        candidates.append(0.0)

    # Stationary point in the upper clipped region, where y^*(x)=R_y 1.
    if -problem.R_x <= 0.0 <= problem.R_x and 0.0 >= upper_kink:
        candidates.append(0.0)

    # Stationary point in the interior response region.
    numerator = (
        problem.gamma
        * problem.k
        * (d * problem.Q_des - problem.k * problem.a)
    )
    denominator = (
        problem.mu_x * d**2 + problem.gamma * problem.k**2
    )
    x_interior = numerator / denominator
    if (
        -problem.R_x <= x_interior <= problem.R_x
        and lower_kink <= x_interior <= upper_kink
    ):
        candidates.append(float(x_interior))

    # Remove numerical duplicates, then evaluate all candidates.
    unique_candidates = sorted(set(round(x, 14) for x in candidates))
    values = [(x, upper_objective(x, problem)) for x in unique_candidates]
    x_star, F_star = min(values, key=lambda pair: pair[1])
    return float(x_star), float(F_star)


# ---------------------------------------------------------------------------
# Surrogate Lagrangian used by Algorithm 1
# ---------------------------------------------------------------------------

def mixed_follower_costs(
    x: float,
    y: Array,
    z: Array,
    problem: CournotProblem,
) -> Array:
    r"""
    Return g_i(x, y_i, z_{-i}) for all i.

    For follower i, the mixed total quantity is
        Q_i = y_i + sum_{j != i} z_j.
    """
    y = np.asarray(y, dtype=float)
    z = np.asarray(z, dtype=float)
    Q_z = float(np.sum(z))
    Q_mixed = y + (Q_z - z)
    return (
        0.5 * problem.c * y**2
        - (problem.a + x - problem.b * Q_mixed) * y
    )


def approximate_lagrangian(
    x: float,
    y: Array,
    z: Array,
    lam: float,
    problem: CournotProblem,
) -> float:
    r"""
    Evaluate
        L_tilde_lambda(x,y,z)
        = f(x,y) + lambda sum_i [g_i(x,y_i,z_-i)-g_i(x,z)].
    """
    penalty = np.sum(
        mixed_follower_costs(x, y, z, problem)
        - follower_costs(x, z, problem)
    )
    return float(leader_cost(x, y, problem) + lam * penalty)


def grad_y_approximate_lagrangian(
    x: float,
    y: Array,
    z: Array,
    lam: float,
    problem: CournotProblem,
) -> Array:
    r"""
    Gradient of L_tilde with respect to y.

    Its i-th coordinate is
        gamma(Q_y-Q_des)
        + lambda[(c+2b)y_i + b sum_{j != i}z_j - (a+x)].
    """
    y = np.asarray(y, dtype=float)
    z = np.asarray(z, dtype=float)
    Q_y = float(np.sum(y))
    Q_z = float(np.sum(z))

    leader_term = problem.gamma * (Q_y - problem.Q_des)
    follower_term = (
        (problem.c + 2.0 * problem.b) * y
        + problem.b * (Q_z - z)
        - (problem.a + x)
    )
    return leader_term + lam * follower_term


def grad_x_approximate_lagrangian(
    x: float,
    y: Array,
    z: Array,
    lam: float,
    problem: CournotProblem,
) -> float:
    r"""
    Partial x-gradient used in the leader update.

    Because partial_x g_i(x,y) = -y_i,
        partial_x L_tilde
        = mu_x x + lambda(sum_i z_i - sum_i y_i).
    """
    return float(
        problem.mu_x * x
        + lam * (float(np.sum(z)) - float(np.sum(y)))
    )


# ---------------------------------------------------------------------------
# Inner solver for line 3: Tseng/Korpelevich extragradient
# ---------------------------------------------------------------------------

def tseng_extragradient(
    x: float,
    z0: Array,
    problem: CournotProblem,
    num_steps: int,
    step_size: Optional[float] = None,
    return_trace: bool = False,
) -> Tuple[Array, Optional[List[Dict[str, float]]]]:
    r"""
    Apply projected extragradient to VI(Y,V(x,.)).

    One inner iteration is
        z_half = P_Y[z - alpha V(x,z)],
        z_next = P_Y[z - alpha V(x,z_half)].

    Tseng's paper studies the map
        T(z)=P_Y[z-alpha V(P_Y[z-alpha V(z)])].
    For a Lipschitz constant L_V, alpha < 1/L_V is the standard safe range.
    """
    if num_steps < 0:
        raise ValueError("num_steps must be nonnegative.")

    L_V = problem.follower_lipschitz_constant
    alpha = 0.9 / L_V if step_size is None else float(step_size)
    if not 0.0 < alpha < 1.0 / L_V:
        raise ValueError(
            f"Extragradient step must satisfy 0 < alpha < 1/L_V={1.0/L_V:.6g}."
        )

    z = project_y(np.asarray(z0, dtype=float), problem)
    trace: Optional[List[Dict[str, float]]] = [] if return_trace else None
    y_star = exact_follower_equilibrium(x, problem) if return_trace else None

    for m in range(num_steps):
        z_half = project_y(z - alpha * follower_operator(x, z, problem), problem)
        z = project_y(
            z - alpha * follower_operator(x, z_half, problem),
            problem,
        )

        if trace is not None and y_star is not None:
            residual = np.linalg.norm(
                z - project_y(
                    z - alpha * follower_operator(x, z, problem),
                    problem,
                )
            )
            trace.append(
                {
                    "inner_iter": float(m + 1),
                    "squared_error": float(np.linalg.norm(z - y_star) ** 2),
                    "projection_residual": float(residual),
                }
            )

    return z, trace


# ---------------------------------------------------------------------------
# Inner solver for line 4/5: projected gradient descent in y
# ---------------------------------------------------------------------------

def y_gradient_descent(
    x: float,
    y0: Array,
    z: Array,
    lam: float,
    problem: CournotProblem,
    num_steps: int,
    step_size: Optional[float] = None,
    return_trace: bool = False,
) -> Tuple[Array, Optional[List[Dict[str, float]]]]:
    r"""
    Minimize y -> L_tilde_lambda(x,y,z) over [0,R_y]^k by projected GD.

    For this quadratic toy instance,
        Hessian_y L_tilde
        = lambda(c+2b)I + gamma 11^T,
    so a valid smoothness constant is
        L_y = lambda(c+2b) + gamma k.
    The standard choice alpha_y = 1/L_y is used by default.
    """
    if num_steps < 0:
        raise ValueError("num_steps must be nonnegative.")
    if lam <= 0:
        raise ValueError("lam must be positive.")

    L_y = lam * (problem.c + 2.0 * problem.b) + problem.gamma * problem.k
    alpha = 1.0 / L_y if step_size is None else float(step_size)
    if not 0.0 < alpha <= 1.0 / L_y:
        raise ValueError(
            f"GD step must satisfy 0 < alpha <= 1/L_y={1.0/L_y:.6g}."
        )

    y = project_y(np.asarray(y0, dtype=float), problem)
    trace: Optional[List[Dict[str, float]]] = [] if return_trace else None

    for m in range(num_steps):
        gradient = grad_y_approximate_lagrangian(x, y, z, lam, problem)
        y = project_y(y - alpha * gradient, problem)

        if trace is not None:
            next_gradient = grad_y_approximate_lagrangian(
                x, y, z, lam, problem
            )
            gradient_mapping = (
                y
                - project_y(y - alpha * next_gradient, problem)
            ) / alpha
            trace.append(
                {
                    "inner_iter": float(m + 1),
                    "lagrangian": approximate_lagrangian(
                        x, y, z, lam, problem
                    ),
                    "gradient_mapping_norm": float(
                        np.linalg.norm(gradient_mapping)
                    ),
                }
            )

    return y, trace


# ---------------------------------------------------------------------------
# Schedules and the complete outer algorithm
# ---------------------------------------------------------------------------

def lambda_schedule(t: int, config: AlgorithmConfig) -> float:
    """lambda_t = lambda0 sqrt(t+1), with zero-based t."""
    return config.lambda0 * math.sqrt(t + 1.0)


def logarithmic_inner_steps(
    t: int,
    base: int,
    coefficient: float,
) -> int:
    """base + ceil(coefficient log(1+t+1)), with zero-based t."""
    return int(base + math.ceil(coefficient * math.log1p(t + 1.0)))


def inner_step_counts(
    t: int,
    config: AlgorithmConfig,
) -> Tuple[int, int]:
    """Return (M_z,t, M_y,t)."""
    Mz = (
        config.fixed_Mz
        if config.fixed_Mz is not None
        else logarithmic_inner_steps(
            t, config.mz_base, config.mz_log_coefficient
        )
    )
    My = (
        config.fixed_My
        if config.fixed_My is not None
        else logarithmic_inner_steps(
            t, config.my_base, config.my_log_coefficient
        )
    )
    return int(Mz), int(My)


def run_algorithm1(
    problem: CournotProblem,
    config: AlgorithmConfig,
) -> Tuple[float, float, float, Array, Array, List[Dict[str, float]]]:
    """Run the deterministic version of Algorithm 1."""

    problem.validate()
    config.validate()

    eta_x = (
        config.eta_x
        if config.eta_x is not None
        else config.eta_x_fraction / problem.interior_upper_smoothness
    )

    x_star, F_star = exact_stackelberg_solution(problem)

    x = project_x(config.x0, problem)
    y = project_y(np.full(problem.k, config.y0_value), problem)
    z = project_y(np.full(problem.k, config.z0_value), problem)

    history: List[Dict[str, float]] = []

    for t in range(config.T):
        iteration_start = time.perf_counter()
        x_before = x
        lam = lambda_schedule(t, config)
        Mz, My = inner_step_counts(t, config)

        # Line 3: approximate y^*(x_t) by extragradient.
        z, _ = tseng_extragradient(
            x=x_before,
            z0=z,
            problem=problem,
            num_steps=Mz,
            step_size=(
                config.extragradient_fraction
                / problem.follower_lipschitz_constant
            ),
            return_trace=False,
        )

        # Line 4/5: approximately minimize L_tilde in y.
        y, _ = y_gradient_descent(
            x=x_before,
            y0=y,
            z=z,
            lam=lam,
            problem=problem,
            num_steps=My,
            return_trace=False,
        )

        # Line 7: projected leader update because X is bounded.
        surrogate_gradient = grad_x_approximate_lagrangian(
            x_before, y, z, lam, problem
        )
        x = project_x(x_before - eta_x * surrogate_gradient, problem)

        # Diagnostics at x_t and x_{t+1}.
        exact_z_at_x_before = exact_follower_equilibrium(
            x_before, problem
        )
        z_error = float(np.linalg.norm(z - exact_z_at_x_before))
        alpha_z = (
            config.extragradient_fraction
            / problem.follower_lipschitz_constant
        )
        z_projection_residual = float(
            np.linalg.norm(
                z
                - project_y(
                    z - alpha_z * follower_operator(x_before, z, problem),
                    problem,
                )
            )
        )

        L_y = lam * (problem.c + 2.0 * problem.b) + problem.gamma * problem.k
        alpha_y = 1.0 / L_y
        y_gradient = grad_y_approximate_lagrangian(
            x_before, y, z, lam, problem
        )
        y_gradient_mapping = (
            y - project_y(y - alpha_y * y_gradient, problem)
        ) / alpha_y

        true_gradient = upper_gradient(x_before, problem)
        F_before = upper_objective(x_before, problem)
        F_after = upper_objective(x, problem)

        iteration_seconds = time.perf_counter() - iteration_start

        history.append(
            {
                "outer_iter": float(t + 1),
                "iteration_seconds": float(iteration_seconds),
                "lambda": float(lam),
                "M_z": float(Mz),
                "M_y": float(My),
                "eta_x": float(eta_x),
                "x_before": float(x_before),
                "x_after": float(x),
                "Q_z": float(np.sum(z)),
                "Q_y": float(np.sum(y)),
                "F_before": float(F_before),
                "F_after": float(F_after),
                "optimality_gap_after": float(abs(F_after - F_star)),
                "surrogate_gradient": float(surrogate_gradient),
                "true_upper_gradient": float(true_gradient),
                "hypergradient_error": float(
                    abs(surrogate_gradient - true_gradient)
                ),
                "z_error": z_error,
                "z_projection_residual": z_projection_residual,
                "y_gradient_mapping_norm": float(
                    np.linalg.norm(y_gradient_mapping)
                ),
            }
        )

    # Return plain values so the caller can unpack them directly.
    return x_star, F_star, x, y, z, history


# ---------------------------------------------------------------------------
# Experiment diagnostics and output
# ---------------------------------------------------------------------------

def run_tseng_contraction_experiment(
    problem: CournotProblem,
    x: float = 0.0,
    num_steps: int = 40,
    seed: int = 7,
) -> List[Dict[str, float]]:
    """
    Hold x fixed and verify the inner extragradient error decays geometrically.
    """
    rng = np.random.default_rng(seed)
    # A non-symmetric initial vector makes the vector VI behavior visible.
    z0 = rng.uniform(0.0, min(problem.R_y, 1.0), size=problem.k)
    _, trace = tseng_extragradient(
        x=x,
        z0=z0,
        problem=problem,
        num_steps=num_steps,
        step_size=0.9 / problem.follower_lipschitz_constant,
        return_trace=True,
    )
    return [] if trace is None else trace


def save_history_csv(
    history: Sequence[Dict[str, float]],
    path: Path,
) -> None:
    """Save the outer-iteration diagnostics."""
    if not history:
        raise ValueError("history is empty.")
    path.parent.mkdir(parents=True, exist_ok=True)
    with path.open("w", newline="") as file:
        writer = csv.DictWriter(file, fieldnames=list(history[0].keys()))
        writer.writeheader()
        writer.writerows(history)


def save_trace_csv(
    trace: Sequence[Dict[str, float]],
    path: Path,
) -> None:
    """Save an inner-solver diagnostic trace."""
    if not trace:
        raise ValueError("trace is empty.")
    path.parent.mkdir(parents=True, exist_ok=True)
    with path.open("w", newline="") as file:
        writer = csv.DictWriter(file, fieldnames=list(trace[0].keys()))
        writer.writeheader()
        writer.writerows(trace)




def display_experiment_plots(
    problem: CournotProblem,
    x_star: float,
    history: List[Dict[str, float]],
) -> None:
    """Display the four requested diagrams from plain input values."""
    import matplotlib.pyplot as plt

    iterations = np.array([int(row["outer_iter"]) for row in history])
    gaps = np.array([row["optimality_gap_after"] for row in history])
    x_path = np.array([row["x_after"] for row in history])
    gradient_norms = np.array(
        [abs(upper_gradient(x, problem)) for x in x_path]
    )
    iteration_times = np.array(
        [row["iteration_seconds"] for row in history]
    )

    plt.figure(figsize=(8, 5.5))
    plt.plot(iterations, gaps, label="First-order method")
    plt.plot(
        iterations,
        1.0 / iterations**2,
        "--",
        label=r"Reference: $1/t^2$",
    )
    plt.xlabel("t")
    plt.ylabel(r"$|F(x_t)-F(x^\star)|$")
    plt.title(r"Optimality gap vs. $1/t^2$")
    plt.legend()
    plt.grid(True, alpha=0.35)
    plt.tight_layout()
    plt.show()

    plt.figure(figsize=(8, 5.5))
    plt.plot(iterations, gradient_norms)
    plt.xlabel("t")
    plt.ylabel(r"$|\nabla F(x_t)|$")
    plt.title("Gradient")
    plt.grid(True, alpha=0.35)
    plt.tight_layout()
    plt.show()

    plt.figure(figsize=(8, 5.5))
    plt.plot(iterations, x_path, label=r"$x_t$")
    plt.axhline(x_star, linestyle="--", label=r"$x^\star$")
    plt.xlabel("t")
    plt.ylabel("Leader")
    plt.title("Iterate")
    plt.legend()
    plt.grid(True, alpha=0.35)
    plt.tight_layout()
    plt.show()

    plt.figure(figsize=(8, 5.5))
    plt.plot(iterations, iteration_times)
    plt.xlabel("Iteration t")
    plt.ylabel("Time per iteration (seconds)")
    plt.title("Fully First-order Method")
    plt.grid(True, alpha=0.35)
    plt.tight_layout()
    plt.show()


def print_summary(
    problem: CournotProblem,
    config: AlgorithmConfig,
    x_star: float,
    F_star: float,
    final_x: float,
    history: List[Dict[str, float]],
) -> Tuple[
    float,
    float,
    float,
    float,
    float,
    float,
    float,
    Optional[int],
    Optional[float],
]:
    """
    Print the requested statistics and return them as plain values.

    Returned order:
        final_F,
        final_gap,
        final_gradient_norm,
        minimum_gap,
        average_time,
        total_time,
        convergence_tolerance,
        convergence_iteration,
        time_to_converge.
    """
    tolerance = config.convergence_gap_tol

    final_F = upper_objective(final_x, problem)
    final_gap = abs(final_F - F_star)
    final_gradient_norm = abs(upper_gradient(final_x, problem))
    minimum_gap = min(row["optimality_gap_after"] for row in history)
    average_time = float(
        np.mean([row["iteration_seconds"] for row in history])
    )
    total_time = float(
        np.sum([row["iteration_seconds"] for row in history])
    )

    cumulative_time = 0.0
    convergence_iteration: Optional[int] = None
    time_to_converge: Optional[float] = None
    for row in history:
        cumulative_time += row["iteration_seconds"]
        if row["optimality_gap_after"] <= tolerance:
            convergence_iteration = int(row["outer_iter"])
            time_to_converge = cumulative_time
            break

    print("\nFully First-order Method: summary")
    print("-" * 52)
    print(f"x*:                         {x_star:.10f}")
    print(f"F(x*):                      {F_star:.10e}")
    print(f"Final x:                    {final_x:.10f}")
    print(f"Final F(x):                 {final_F:.10e}")
    print(f"Final optimality gap:       {final_gap:.10e}")
    print(f"|∇F(final x)|:              {final_gradient_norm:.10e}")
    print(f"Minimum optimality gap:     {minimum_gap:.10e}")
    print(f"Average time / iteration:   {average_time:.6e} seconds")
    print(f"Total runtime:              {total_time:.6e} seconds")
    print(f"Convergence tolerance:      {tolerance:.1e}")

    if time_to_converge is None:
        print("Convergence iteration:      not reached")
        print("Time to converge:           not reached")
    else:
        print(f"Convergence iteration:      {convergence_iteration}")
        print(f"Time to converge:           {time_to_converge:.6e} seconds")

    return (
        final_F,
        final_gap,
        final_gradient_norm,
        minimum_gap,
        average_time,
        total_time,
        tolerance,
        convergence_iteration,
        time_to_converge,
    )


def run_default_experiment() -> Tuple[
    float,
    float,
    float,
    Array,
    Array,
    List[Dict[str, float]],
    float,
    float,
    float,
    float,
    float,
    float,
    float,
    Optional[int],
    Optional[float],
]:
    """
    Run the default experiment and return plain values.

    Example
    -------
    (
        x_star,
        F_star,
        final_x,
        final_y,
        final_z,
        history,
        final_F,
        final_gap,
        final_gradient_norm,
        minimum_gap,
        average_time,
        total_time,
        convergence_tolerance,
        convergence_iteration,
        time_to_converge,
    ) = run_default_experiment()
    """
    problem = CournotProblem()
    config = AlgorithmConfig()

    (
        x_star,
        F_star,
        final_x,
        final_y,
        final_z,
        history,
    ) = run_algorithm1(problem, config)

    (
        final_F,
        final_gap,
        final_gradient_norm,
        minimum_gap,
        average_time,
        total_time,
        convergence_tolerance,
        convergence_iteration,
        time_to_converge,
    ) = print_summary(
        problem=problem,
        config=config,
        x_star=x_star,
        F_star=F_star,
        final_x=final_x,
        history=history,
    )

    display_experiment_plots(
        problem=problem,
        x_star=x_star,
        history=history,
    )

    return (
        x_star,
        F_star,
        final_x,
        final_y,
        final_z,
        history,
        final_F,
        final_gap,
        final_gradient_norm,
        minimum_gap,
        average_time,
        total_time,
        convergence_tolerance,
        convergence_iteration,
        time_to_converge,
    )


if __name__ == "__main__":
    run_default_experiment()
```

## Zero-th order method


```python

import numpy as np
import matplotlib.pyplot as plt
import math
import time

# =========================
# 1) Cournot bilevel instance (same setup as we've been using)
# =========================
a=12; b=0; c=2000; k=2000
Rx=100; Ry=1000
mu_x=10; gamma=1; Qdes=10
den=c+b*(k+1)

def proj_x(x: float) -> float:
    return float(np.clip(x, -Rx, Rx))

def proj_y(y: np.ndarray) -> np.ndarray:
    return np.clip(y, 0.0, Ry)

def Q(y: np.ndarray) -> float:
    return float(np.sum(y))

def y_star(x: float) -> np.ndarray:
    """True follower equilibrium (closed form symmetric equilibrium for this Cournot)."""
    denom = c + b * (k + 1.0)
    yi = np.clip((a + x) / denom, 0.0, Ry)
    return yi * np.ones(k)

def f(x: float, y: np.ndarray) -> float:
    """Leader objective at (x,y)."""
    return 0.5 * mu_x * x * x + 0.5 * gamma * (Q(y) - Qdes) ** 2

def F(x: float) -> float:
    """True bilevel objective F(x)=f(x,y*(x))."""
    return f(x, y_star(x))

# Closed-form minimizer of F(x) for this instance (quadratic in x), projected to [-Rx,Rx]
denom = c + b * (k + 1.0)
A = mu_x + gamma * (k / denom) ** 2
B = gamma * (k / denom) * (k * a / denom - Qdes)
x_star = proj_x(-B / A)
F_star = F(x_star)

print("Problem parameters:")
print(f"a={a}, b={b}, c={c}, k={k}, Rx={Rx}, Ry={Ry}, mu_x={mu_x}, gamma={gamma}, Qdes={Qdes}")
print(f"x* ≈ {x_star:.6f},  F(x*) ≈ {F_star:.6f}")

# =========================
# 2) Inner solver: ApproxNE via Gauss–Seidel best response (works for b>0)
# =========================
def approx_NE(x: float, K: int = 40, y0: np.ndarray | None = None) -> np.ndarray:
    """
    Approximate Nash equilibrium for followers given x by best-response iteration:
      y_i <- proj( (a + x - b*sum_{j≠i} y_j) / (c + 2b), [0,Ry] ).
    """
    if y0 is None:
        y = np.zeros(k, dtype=float)
    else:
        y = y0.astype(float).copy()

    denom_br = c + 2.0 * b
    for _ in range(K):
        S = float(np.sum(y))
        for i in range(k):
            S_minus = S - float(y[i])
            yi = (a + x - b * S_minus) / denom_br
            yi = float(np.clip(yi, 0.0, Ry))
            # update running sum efficiently
            S = S_minus + yi
            y[i] = yi
    return y

# =========================
# 3) Gaussian Process surrogate (RBF) + LCB acquisition (BO-style)
#    Tuned for speed:
#      - small candidate grid
#      - cap GP memory to a small window
#      - vectorized kernel computations
#      - larger jitter/noise for numerical stability
# =========================
def rbf_kernel_vec(X: np.ndarray, Y: np.ndarray, ell: float, var: float) -> np.ndarray:
    """Vectorized RBF kernel between 1D arrays X (n,), Y (m,)."""
    X = X[:, None]
    Y = Y[None, :]
    d2 = (X - Y) ** 2
    return var * np.exp(-0.5 * d2 / (ell ** 2))

def gp_posterior(X: np.ndarray, y: np.ndarray, Xtest: np.ndarray,
                 noise: float, ell: float, var: float):
    """
    GP posterior mean/std for 1D inputs.
    Returns (mu, std) at Xtest.
    """
    # Standardize observations for stability
    y_mean = float(y.mean())
    y_std = float(y.std())
    if y_std < 1e-12:
        y_std = 1.0
    ys = (y - y_mean) / y_std

    K = rbf_kernel_vec(X, X, ell, var) + (noise + 1e-10) * np.eye(len(X))
    Ks = rbf_kernel_vec(X, Xtest, ell, var)

    # Cholesky
    L = np.linalg.cholesky(K)
    alpha = np.linalg.solve(L.T, np.linalg.solve(L, ys))

    mu_s = Ks.T @ alpha
    v = np.linalg.solve(L, Ks)
    var_s = np.maximum(var - np.sum(v * v, axis=0), 1e-12)
    std_s = np.sqrt(var_s)

    mu = mu_s * y_std + y_mean
    std = std_s * y_std
    return mu, std

# =========================
# 4) Maddux Algorithm 1 loop + suffix-average plot
# =========================
T = 600                # Keep moderate so notebook runs fast
Nwarm = 25              # warm-start random evaluations

grid = np.linspace(-Rx, Rx, 121)   # candidate set (smaller -> faster)

max_n = 15              # cap GP memory tightly for speed (sliding window)
noise = 2e-3            # GP observation noise / jitter (bigger -> more stable)
ell = 3.0               # kernel lengthscale
var = 1.0               # kernel variance

beta0 = 8.0
def beta_t(t: int) -> float:
    # Decaying exploration weight to encourage convergence
    return beta0 / math.sqrt(max(t, 1))

rng = np.random.default_rng(0)

X_hist = []     # x points used by GP
y_hist = []     # observed objectives for GP (f(x, y_approx))
xs = []         # algorithm-chosen x_t
gaps = []       # |F(x_t)-F(x*)|
y_prev = None   # warm start for inner NE solver
time_elapsed = np.zeros(T)

for t in range(1, T + 1):
    # Choose x_t
    start_time = time.time()
    if t <= Nwarm:
        x_t = float(rng.uniform(-Rx, Rx))
    else:
        X_np = np.asarray(X_hist, dtype=float)
        y_np = np.asarray(y_hist, dtype=float)
        mu, std = gp_posterior(X_np, y_np, grid, noise=noise, ell=ell, var=var)

        # LCB for minimization
        acq = mu - beta_t(t) * std
        x_t = proj_x(float(grid[int(np.argmin(acq))]))

    # Inner solve accuracy grows slowly with t (helps stability/convergence)
    #K_inner = int(30 + 10 * math.log(t + 1))
    y_t = approx_NE(x_t, y0=y_prev)
    y_prev = y_t

    # Observe bandit feedback
    J_obs = f(x_t, y_t)
    X_hist.append(x_t)
    y_hist.append(J_obs)
    end_time = time.time()
    time_elapsed[t-1] = end_time - start_time

    # Record metrics using the true bilevel objective (since we know y*(x))
    xs.append(x_t)
    gaps.append(abs(F(x_t) - F_star))

xs = np.asarray(xs, dtype=float)
gaps_0th = np.asarray(gaps, dtype=float)



# =========================
# 5) Plots
# =========================
plt.figure(figsize=(7.2, 4.2))
plt.semilogy(gaps_0th, linewidth=2)
plt.xlabel("iteration t")
plt.ylabel(r"$_{200}$ $|F(x_t)-F(x^\star)|$")
plt.title("Maddux Algorithm 1: optimality gap")
plt.grid(True, which="both", alpha=0.3)
plt.show()

plt.figure(figsize=(7.2, 3.6))
plt.plot(xs, alpha=0.9)
plt.axhline(x_star, linestyle="--", label=r"$x^\star$")
plt.xlabel("iteration t")
plt.ylabel(r"$x_t$")
plt.title("Maddux Algorithm 1: chosen $x_t$")
plt.grid(True, alpha=0.3)
plt.legend()
plt.show()

plt.figure(figsize=(7.2, 3.6))
plt.plot(time_elapsed[:T])
plt.xlabel("iteration t")
plt.ylabel("Time per iteration")
plt.title("Maddux Algorithm 1: time ")
plt.grid(True, alpha=0.3)

plt.show()

# =========================
# 6) Summary
# =========================
best_idx = int(np.argmin(gaps_0th))
print("Summary:")
print (f"Time per iteration: {time_elapsed.mean()}")
print(f"  Best raw gap: {gaps_0th[best_idx]:.3e} at t={best_idx+1}, x={xs[best_idx]:.6f}")
print(f"  Final regret (W=200): {gaps_0th[-1]:.3e}")
print(f"Time to converges: {sum(time_elapsed[0:600])}")
```

## Second-order method

```python
import numpy as np
import matplotlib.pyplot as plt
import time

# =========================
# Cournot instance
# =========================


a=12; b=0; c=2000; k=2000
Rx=100; Ry=1000
mu_x=10; gamma=1; Q_des=10


def Q(y):
    return float(np.sum(y))

# =========================
# Oracle: black-box access to follower equilibrium y*(pi)
# (internally uses the known closed-form, but the algorithm only calls this function)
# =========================
def oracle_y_star(pi):
    denom = c + b*(k+1.0)
    yi = (a + pi)/denom
    yi = float(np.clip(yi, 0.0, Ry))
    return yi*np.ones(k, dtype=float)

# =========================
# Cournot leader objective f(pi,y) and bilevel objective F(pi)=f(pi,y*(pi))
# =========================
def f_cournot(pi, y):
    return 0.5*mu_x*pi*pi + 0.5*gamma*(Q(y)-Q_des)**2

def F(pi):
    y = oracle_y_star(pi)
    return f_cournot(pi, y)

# =========================
# Wang paper's f and g
# You asked: MINIMIZE F(pi) <=> MAXIMIZE -F(pi)
# Eq.(10) uses -E[f_paper], so set f_paper := - f_cournot,
# then -E[f_paper] = E[f_cournot] = F.
# =========================
def f_paper(pi, ystar):
    return -f_cournot(pi, ystar)

# In your screenshot Cournot has NO leader constraints, so g ≡ 0.
# (Then lambda/slack updates do nothing, but the implementation matches Alg.1 structure.)
def g_paper(pi, ystar):
    return 0.0

# =========================
# Eq.(5): compute dy*/dpi as (dF/dy)^{-1} * (-dF/dpi) for follower stationarity system.
#
# Cournot stationarity residual (interior) is:
#   F_i(y,pi) = (c+b) y_i + b(1^T y) - (a+pi) = 0
# Jacobian wrt y: J = (c+b)I + b 11^T
# dF/dpi = -1 vector, so RHS = 1 vector.
#
# Need v = J^{-1} 1.
# =========================
# --- global cache for J (optional but recommended) ---
_J_cache = None

def dy_dpi_via_eq5_solve(pi: float) -> np.ndarray:
    """
    Compute dy*/dpi via Eq.(5) by explicitly solving the linear system
        J v = rhs
    where
        J = (c+b) I + b 11^T
        rhs = 1

    This is mathematically equivalent to using J^{-1},
    but numerically stable and fast.
    """
    global _J_cache

    denom = c + b*(k+1.0)
    yi_unproj = (a + pi) / denom

    # If projection is active, equilibrium is locally flat
    if yi_unproj <= 0.0 or yi_unproj >= Ry:
        return np.zeros(k, dtype=float)

    # Build J once (it does not depend on pi in the interior)
    if _J_cache is None:
        I = np.eye(k, dtype=float)
        ones = np.ones((k, 1), dtype=float)
        _J_cache = (c + b) * I + b * (ones @ ones.T)

    rhs = np.ones(k, dtype=float)
    return np.linalg.solve(_J_cache, rhs)


# =========================
# Algorithm 1: Augmented Lagrangian method (Wang et al.)
# For your unconstrained Cournot screenshot, g≡0 so lambda/s are inert;
# the pi-update becomes gradient descent on F(pi), computed via oracle + Eq.(5).
# =========================
def run_wang_alg1(T=500, pi0=0.0, lr=0.01, mu_pen=1.0, K=100, lam0=0.0, s0=0.0):
    pi = pi0
    lam = float(lam0)
    s = max(float(s0), 0.0)

    pis = np.zeros(T+1)
    Fs  = np.zeros(T+1)
    gradabs = np.zeros(T+1)

    time_elapsed = np.zeros(T+1)

    for t in range(T+1):

        start_time = time.perf_counter()
        pis[t] = pi

        # monitor true gradient of F using oracle + Eq.(5)
        ystar = oracle_y_star(pi)
        Fs[t] = f_cournot(pi, ystar)
        dy = dy_dpi_via_eq5_solve(pi)

        dQdpi = float(np.sum(dy))

        # dF/dpi = mu_x*pi + gamma*(Q(y*)-Q_des)* dQ/dpi
        gradF = mu_x*pi + gamma*(Q(ystar)-Q_des)*dQdpi
        gradabs[t] = abs(gradF)

        if t == T:
            break

        # ---- Alg.1 line 4: sampled gradient of L by sampling x*~O(pi)
        # Here x* is y* from oracle, and f_paper is -f_cournot so -E[f_paper]=F
        # With g≡0, L = F(pi) + lam*(0+s) + (mu/2)(0+s)^2 = F(pi) + lam*s + (mu/2)s^2
        # pi-gradient is just gradF (since g does not depend on pi in this unconstrained example).
        dL_dpi = gradF

        # s-gradient: dL/ds = lam + mu*s  (since g≡0)
        dL_ds = lam + mu_pen*s

        # ---- Alg.1 line 5: update pi and s
        pi = (pi - lr*dL_dpi)
        s = max(s - lr*dL_ds, 0.0)


        # ---- Alg.1 line 6-7: lambda update every K steps (MATCH pseudo-code)
        if (t+1) % K == 0:
            gval = g_paper(pi, oracle_y_star(pi))  # 0.0 here
            lam = lam - mu_pen*(gval + s)
        end_time = time.perf_counter()
        time_elapsed[t] = end_time - start_time

    return pis, Fs, gradabs, time_elapsed

# =========================
# Run + plots
# =========================
T=600
pis, Fs, gradabs, time_elapsed = run_wang_alg1(T=T, pi0=0.0, lr=0.01, mu_pen=1.0, K=100, lam0=0.0, s0=0.0)

# Compute reference optimum by 1D golden search on F (for plotting gap)
def golden_min(f, a0, b0, tol=1e-12, maxit=4000):
    import math
    gr=(math.sqrt(5)-1)/2
    a_=a0; b_=b0
    c_=b_-gr*(b_-a_); d_=a_+gr*(b_-a_)
    fc=f(c_); fd=f(d_)
    for _ in range(maxit):
        if abs(b_-a_)<tol: break
        if fc<fd:
            b_=d_; d_=c_; fd=fc
            c_=b_-gr*(b_-a_); fc=f(c_)
        else:
            a_=c_; c_=d_; fc=fd
            d_=a_+gr*(b_-a_); fd=f(d_)
    x_=(a_+b_)/2
    return x_, f(x_)

pi_star, F_star = golden_min(F, -Rx, Rx)
gaps_2nd = np.abs(Fs - F_star)

print("pi* =", pi_star, "F* =", F_star)
print("pi_T =", pis[-1], "gap_T =", gaps_2nd[-1], "grad_T =", gradabs[-1])
print("Average time per iteration =", np.mean(time_elapsed))
print("Time to converges= ", sum(time_elapsed[0:300]))

plt.figure(figsize=(7,4))
plt.semilogy(np.maximum(gradabs, 1e-18))
plt.xlabel("iteration t"); plt.ylabel(r"$|\nabla F(\pi_t)|$")
plt.title("Wang et al Alg.1: gradient")
plt.grid(True, alpha=0.3); plt.tight_layout()
plt.show()

plt.figure(figsize=(7,4))
plt.semilogy(np.maximum(gaps_2nd, 1e-18))
plt.xlabel("iteration t"); plt.ylabel(r"$|F(\pi_t)-F(\pi^\star)|$")
plt.title("Wang et al Alg.1 (oracle + Eq.(5) inverse): optimality gap")
plt.grid(True, alpha=0.3); plt.tight_layout()
plt.show()


plt.figure()
plt.semilogy(time_elapsed[:T])
plt.xlabel("iteration t")
plt.ylabel("Time per iteration")
plt.title("Wang et al Alg.1: time ")
plt.grid(True, alpha=0.3)

plt.show()
```
